```python
import numpy as np
import pandas as pd
from IPython.display import display

# Fungsi Aktivasi Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Input dan Label
X = np.array([0.5, 0.3])
y = 1
learning_rate = 0.1

# Inisialisasi Bobot Awal dan Bias
W1 = np.array([[0.4, 0.2],  
               [0.3, 0.6]]) 
b1 = np.array([0.1, -0.1])

W2 = np.array([0.5, -0.4])  
b2 = 0.2

print("=== SETUP AWAL: BOBOT HIDDEN LAYER (W1) ===")
display(pd.DataFrame(W1, columns=["Bobot dari x1", "Bobot dari x2"], index=["Neuron H1_1", "Neuron H1_2"]))
```

# Langkah 1: Menghitung Forward Propagation

```python

# 1. Menghitung z dan h di Hidden Layer
z1 = np.dot(W1, X) + b1
h1 = sigmoid(z1)
print(f"z11: {z1[0]:.4f}, h11: {h1[0]:.4f}")
print(f"z12: {z1[1]:.4f}, h12: {h1[1]:.4f}")

# 2. Menghitung z dan y_hat di Output Layer
z2 = np.dot(W2, h1) + b2
y_hat = sigmoid(z2)
print(f"\nzo: {z2:.4f}, y_hat: {y_hat:.4f}")

# 3. Menghitung Loss (Binary Cross Entropy)
loss = -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
print(f"Loss BCE: {loss:.4f}")

# 4. Prediksi Kelas
pred_class = 1 if y_hat >= 0.5 else 0
print(f"Prediksi Kelas: {pred_class}")
```

# Langkah 2: Menghitung Backward Propagation

```python

# 1. Error Signal Output (delta3)
delta3 = y_hat - y
print(f"Delta 3: {delta3:.4f}")

# 2. Gradien Output Layer
dW2 = delta3 * h1
db2 = delta3
print(f"Gradien W2 (w211, w212): {dW2}")
print(f"Gradien b2: {db2:.4f}")

# 3. Error Signal Hidden Layer (delta1)
delta1 = (delta3 * W2) * (h1 * (1 - h1))
print(f"\nDelta 1 (delta1_1, delta1_2): {delta1}")

# 4. Gradien Hidden Layer
dW1 = np.outer(delta1, X)
db1 = delta1
print(f"Gradien W1:\n{dW1}")
print(f"Gradien b1: {db1}")
```

```python
print("--- UPDATE BOBOT ---")

# 5. Update Bobot
W2_new = W2 - learning_rate * dW2
b2_new = b2 - learning_rate * db2

W1_new = W1 - learning_rate * dW1
b1_new = b1 - learning_rate * db1

print("--- BOBOT BARU SETELAH 1 ITERASI UPDATE ---")
print(f"W2_baru: {W2_new}")
print(f"b2_baru: {b2_new:.4f}")
print(f"W1_baru:\n{W1_new}")
print(f"b1_baru: {b1_new}")
```
